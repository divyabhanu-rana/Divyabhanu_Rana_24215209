```python
import numpy as np
```

```python
# Step Activation Function
def step(x):
  if x >= 0:
    return 1
  else:
    return 0
```

```python
# Training Data (using AND Gate)
X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y = np.array([0, 0, 0, 1])
```

```python
# Initialize weights and bias
weights = np.zeros(2)
bias = 0
learning_rate = 0.1
epochs = 10
```

```python
# Training
for epoch in range(epochs):
    for i in range(len(X)):
        x = X[i]
        target = y[i]

        # Calculate output
        net = np.dot(x, weights) + bias
        output = step(net)

        # Error
        error = target - output

        # Update weights and bias
        weights = weights + learning_rate * error * x
        bias = bias + learning_rate * error
```

```python
# Testing
print("Final Weights:", weights)
print("Final Bias:", bias)

print("\nPredictions:")
for x in X:
    net = np.dot(x, weights) + bias
    prediction = step(net)
    print(x, "->", prediction)
```
