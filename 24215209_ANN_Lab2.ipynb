```python
import numpy as np
```

```python
# Define activation function (in this case, Sigmoid)
def sigmoid(x):
  return 1 / (1 + np.exp(-x))
```

```python
# Input (using AND Gate)
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([0, 0, 0, 1])
```

```python
weights = np.zeros(2)
bias = 0
learning_rate = 0.1
epochs = 100
```

```python
for epoch in range(epochs):
    for i in range(len(X)):
        net = np.dot(X[i], weights) + bias
        output = sigmoid(net)

        error = y[i] - output

        weights += learning_rate * error * X[i]
        bias += learning_rate * error
```

```python
# Testing
print("Predictions:\n")

for x in X:
    net = np.dot(x, weights) + bias
    output = sigmoid(net)
    prediction = 1 if output >= 0.5 else 0

    print("Input:", x)
    print("Output:", round(output, 4))
    print("Prediction:", prediction)
    print()
```
